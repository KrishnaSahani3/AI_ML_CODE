```python
# Selecting data
# COLUMNS VALUES KO ISS TARAH KARTE HAI 
df["city"] # saare city ko ouput de dega

df[["city","aqi"]] # multiple variables ko print karne ke liye double []  lagana important hai

# ROW VALUES KO ISS TARAH
df.loc[0:3 , "city"]  # [] iske andar values row no. ko define karengi (3 inclusive rahega)

# df.iloc[0:3 , "city"]   # error aa jaega because city is string and iloc support numerical values only

# # df.iloc[0:2]  # this id for columns

# # isme index se data lete sakte hai (2 exclusive rahega)

# df.columns  # yaha se pta chala ki city ka serial no 2 then

# df.iloc[0:3 ,2] # ab ye sahi hai

# #   INDIVISUALS CELLS - rows,columns
df.loc[0 ,"aqi"]
df.loc[0:2,["country","city","aqi","longitude","latitude"]]
df.iloc[0:3 ,1:6]

# SINGLE SCALAR VALUE (SINGLE  ROW OR COLUMN)
df.at[0, "city"]
df.iat[0, 2]
## IF ORIGINAL DATA KO  CHANGE KARNE CHAHTE HAI TOH
cities = df["city"]
cities[0] = "New York"
print(cities)


# If original data ko change nhi karna chahte hai to .copy() laga do
cities = df["city"].copy()
cities[0] = "AZAMGARH"
print(cities)


```

```python
## FILTERING OF DATA
# df[ df["aqi"] > 100]
df[(df["aqi"] >100) & (df["temperature"] >30) ].sort_values(["aqi" , "temperature"])

df[(df["aqi"] >100) & (df["temperature"] <30) ].sort_values(["aqi" , "temperature"])

aqi_data =df[df["aqi"] > 100 ] [["city", "aqi" , "temperature"]]
# print(aqi_data)

aqi_data.iloc[0]  # this is for normal indexing
aqi_data.loc[6]  # this is for default indexing  randomly choosen from list

# IF WE WANT TO CORRECT THE INDEXING VALUE THEN 
aqi_data.reset_index(drop= True) 


# QUERY METHOD  cpoy return karta hai

df.query("aqi>100 & temperature>30")


```

```python
# CLEANING DATA - missing  values

df = pd.read_csv("raw_data.csv")

df.isnull()  # jaha par null rahega waha par true dega otherwise false

df.isna() # both are same

df.isnull().sum() # missing value de dega

df.dropna()  # row drop  of empty cells
df.dropna(axis = 1)  # column drop

df.fillna(0)  # fill up with 0 in empty sets we can change 0 to any random value

df.["age"].fillna(df["age"])


```
