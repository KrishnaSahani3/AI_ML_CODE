```python
# Series in pandas
s = pd.Series([1, 2, 3, 4, 5, 18])
print(type(s))

# index
print(s[0])
print(s[1])
print(s[2])
```

```python
# labelled series
s = pd.Series([18, 36, 34], index = ["KRISHNA", "VIRAT KOHLI","ANUSHKA"])
print(s)
print(s["KRISHNA"])
print(s["VIRAT KOHLI"])
print(s["ANUSHKA"])
# print(s[0])
print(s.index) # labellled value de dega
```

```python
### PANDAS IN SERIES
'''
1-series are homogenous
2- vectorized operations
3- handle missing value with NaN
4- mutable values, immuatable size
'''
# vectorized operation
s1 = pd.Series([1, 2, 3, 4, 18])
s2 = pd. Series([45, 23, 17, 12, 18])
print(s1 + s2)

# mutable value but immutable size (size of that sweries is immutable means we can't add new element)
s1[0] = 1000
print(s1)
changed_s1 = s1.drop(0)   # kisi bhi element ko drop ka dega
print(changed_s1)
```
