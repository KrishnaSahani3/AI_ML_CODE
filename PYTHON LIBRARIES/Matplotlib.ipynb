```python
# colors and label
df = pd.read_csv("batsman_season_record.csv")
df
```

```python
plt.bar(df["batsman"] , df["2015"] )
plt.bar(df["batsman"] , df["2016"] )
plt.bar(df["batsman"] , df["2017"] )
```

```python
plt.bar(np.arange(df.shape[0]) - 0.2 , df['2015'] , width = 0.2)
plt.bar(np.arange(df.shape[0])  , df['2016'] , width = 0.2)
plt.bar(np.arange(df.shape[0]) + 0.2, df['2017'] , width = 0.2)

plt.xticks(np.arange(df.shape[0]) , df["batsman"])   # xticks ka mean replacing x axis label to another label but first you have to give previous one then one you want to replace
plt.show()
```

```python
# if x axis name is too big and overlapping
children1 = [10, 20, 30 , 40, 80, 10]
color = ["red red red red" , "blue blue blue blue" , "green" , "orange orange orange orange orange " , 'brown' , 'grey']
plt.bar(color, children1)  
plt.xticks(rotation ="vertical")
plt.show()
```

```python
# Stacked bar chart
plt.bar(df["batsman"] ,df['2015'] , label = '2015')
plt.bar(df["batsman"] ,df['2016'] , bottom = df['2015'] , label = '2016')
plt.bar(df["batsman"] ,df['2017'] , bottom = (df['2015'] + df['2016']), label = '2017')

plt.legend()
plt.show()
```

3D SURFACE PLOTS

```python
x = np.linspace(-10 , 10 , 100)
y = np.linspace(-10 ,10 , 100)
```

```python
xx , yy = np.meshgrid(x,y)
xx.shape
yy.shape
```

```python
z = xx**2 + yy**2
z.shape
```

```python
fig  =plt.figure(figsize = (12,8))

ax = plt.subplot(projection = "3d")
c = ax.plot_surface(xx,yy ,z , cmap = "viridis")
fig.colorbar(c)
```

```python
z = np.sin(xx) + np.cos(yy)
fig  = plt.figure(figsize = (12,8))
ax = plt.subplot(projection = "3d")
p = ax.plot_surface(xx ,yy, z ,cmap = "viridis")
fig.colorbar(p)
```

3D TO 2D


```python
z = xx**2 + yy**2
fig  = plt.figure(figsize = (12,8))
ax = plt.subplot()
p = ax.contour(xx ,yy, z )
fig.colorbar(p)
```

```python
# fill contour by contourf
fig  = plt.figure()
ax = plt.subplot()
p = plt.contourf(xx,yy,z , color = "viridis")
plt.colorbar(p)
```

```python
z = np.sin(xx) + np.cos(yy)
fig = plt.figure()
ax = plt.subplot()
p = plt.contour(xx ,yy , z , color = "viridis")
plt.colorbar(p)
```
