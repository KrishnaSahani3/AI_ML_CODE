```python
import json
```

```python
#Load the data
def load_data(filename):
    with open(filename ,"r") as f:
        data = json.load(f)
    return data
    
```

```python
data = load_data("store_data.json")
print(data)
print(type(data))
```

```python
# Clean & structure data
def clean_data(data):
    text_to_num = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5}
    cleaned_data = []
    unique_users = set()

    for user in data:
        # Clean rating
        raw_rating = str(user["rating"]).strip().lower()
        if raw_rating in text_to_num:
            raw_rating = text_to_num[raw_rating]
        user["rating"] = raw_rating

        # Handle missing age
        raw_age = user.get("age")
        if raw_age is None:
            user["age"] = None

        # De-duplication
        name = user["name"].strip()
        if name in unique_users:
            continue
        unique_users.add(name)

        cleaned_data.append(user)

    return cleaned_data


   
    
```

```python
data = clean_data(data)
```

```python
# Get meaningful insights from data
def get_insight(data):


    # avg rating
    tot_rating = 0
    for user in data:
        tot_rating += float (user["rating"])

    print(f"avg  rating = {tot_rating/len(data)}")

    # percentagw of users with poor rating
    poor_rating = 0
    for user in data:
        if(float(user["rating"]) <3 ):
            poor_rating +=1

        p_rating = poor_rating/len(data)*100
        print(p_rating)


    
            
    
        
```

```python
get_insight(data)
```

```python
# Build recommendation Feature
def get_recommendation(data):
    recommendations = []

    for user in data:
        curr_recomm ={}

        if(float(user["rating"]) >=4):
            curr_recomm["brand"] = "Apple"
        else:
            curr_recomm["brand"] = "Samsung"

        recommendations.append(curr_recomm)
    return recommendations



    
    

    
   


    
```
