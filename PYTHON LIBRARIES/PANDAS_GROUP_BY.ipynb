```python
"""AGGRIGATE OPERATION TAB USR KARTE HAI JAB SPECIFIC COLUMNS KE UPAR
ALAG ALAG OPERATION  PERFORM KARNA HOTA HAI """
# passing dict
genres.agg(
    {
        "Runtime" : "mean",
        "IMDB_Rating" : "mean",
        "No_of_Votes" : "sum",
        "Gross": "sum",
        "Metascore": "sum",
        
    }

)
    
```

```python
"""BY THIS WE CAN FIND MIN , MAX ,SUM  OF EVERY COLUMNS"""

genres.agg(["min", "max", "sum"])
```

```python
"""MERGING BOTH AGG """

genres.agg(
    {
        "Runtime" : ["max","mean"],
        "IMDB_Rating" : "mean",
        "No_of_Votes" : ["mean" ,"sum"],
        "Gross": ["max","sum"],
        "Metascore": "sum",

    }

)
```

```python
"""LOOPING OF GROUPS"""
df  = pd.DataFrame(columns = movies.columns)
for group , data in genres:
    temp = data[data["IMDB_Rating"] == data['IMDB_Rating'].max()]
    df  = pd.concat([df , temp] , ignore_index = True)
df
```

```python
# find number of movies starting with A for each group

def foo(group):
    return group["Series_Title"].str.startswith("A").sum()
```

```python
# find ranking of each movie in the group according to IMDB score 
def rank_movies(group):
    group["genre_rank"] = group["IMDB_Rating"].rank(ascending = False)
    return group

```

```python
genres.apply(rank_movies)
```

```python
# find normalized IMDB rating group wise
def normal(group):
    min_rating = group["IMDB_Rating"].min()
    max_rating = group["IMDB_Rating"].max()
    if max_rating - min_rating == 0:
        print("Can't divide by zero")

    else: 
         group["norm_rating"] = group["IMDB_Rating"] - group["IMDB_Rating"].min()/(max_rating - min_rating)
    return group

genres.apply(normal)
```
