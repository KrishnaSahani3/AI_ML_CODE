# How to create an array from existing data

```python
a  = np.arange(1,11)
a
```

```python
# new array by slicing 
a1  = a[3:8]
a1
```

```python
# by vstack and hstack
arr1 = np.array([[1, 1],
                 [2, 2]])
arr2 = np.array([[3, 3] ,
                 [4, 4]])
# vertical add karna 
np.vstack((arr1 , arr2))
```

```python
# horizontal add karna 
np.hstack((arr1 ,arr2))
```

```python
# we can split an array into several smaller arrays using hsplit or vsplit or (dsplit for 3d)
x = np.arange(1, 25).reshape(2, 12)
x
```

```python
np.hsplit(x , 3)
```

```python
np.vsplit(x ,2)
```
