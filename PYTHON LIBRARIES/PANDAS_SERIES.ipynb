```python
import pandas as pd
```

SERIES FROM LISTS


```python
country = ["India","Pakistan" , "USA", "Srilanka"]
pd.Series(country)
```

```python
# INTEGER
runs = [12, 18, 45, 264, 100]
pd.Series(runs)
```

```python
# custom index
marks = [67, 57 , 89, 100]
subjects = ['maths' , 'english','science' ,'hindi']

pd.Series(marks , index = subjects)
```

```python
# Setting a name
marks = pd.Series(marks, index = subjects , name = 'Krishna ke marks')
marks
```

SERIES FROM DICT


```python
marks = {
    "maths": 67,
    "english": 57,
    "science": 89,
    "hindi":100
    
}
marks_series  = pd.Series(marks , name = "Krishna ke marks ")
marks_series
```

SERIES ATTRIBUTES


```python
# size of series bta dega ki kitne element hai
a = marks_series.size
print(a)

# dtype  data type bta dega
b = marks_series.dtype
print(b)

# name  [topic ke baare mein bta  dega if mention hai toh]
c = marks_series.name
print(c)

# is_unique
d = marks_series.is_unique
print (d)

pd.Series([1, 1, 2, 2, 2, 3, 4, 5 ,56]).is_unique # FALSE DEGA BECAUSE REPEATED NUMBER HAI 


# index 

e = marks_series.index
print(a)
# values

f = marks_series.values
print(f)

# keys() 
k = marks_series.keys()
print(k)
```
