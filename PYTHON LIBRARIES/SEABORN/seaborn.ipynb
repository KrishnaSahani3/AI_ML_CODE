# RELATIONAL PLOT
1. Scatter plot
2. Lineplot

```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
tips = sns.load_dataset("tips")
tips
```

```python
# scatter plot  -> is axis level function

sns.scatterplot(tips, x = "total_bill" , y = "tip" , hue ="sex" , style = "time" , size = "size")
```

```python
# relplot -> figure level -> square shape
sns.relplot(data = tips , x = "total_bill" , y  = "tip" , kind = "scatter" , hue = "sex" , style = "time" , size = "size")
```

```python
# lineplot 
```

```python
gap = px.data.gapminder()
gap
```

```python
temp_df = gap[gap["country"] == "India"]
temp_df.sample(10)
```

```python
# axes lrvel function
sns.lineplot(temp_df , x = "year" , y  = "lifeExp")
```

```python
# figure level function
sns.relplot(temp_df , x = "year" , y = "lifeExp" , kind = "line")
```

# DISTRIBUTION PLOT
1. used for univariate analysis
2. used to find out the distribution
3. range of the observation
4. is the dats bimodal?
5. are there outliers ?

PLOTS UNDER DISTRIBUTION PLOT
1. histplot
2. kdeplot
3. rugplot

```python
# figure level known as displot
# axes level known as histplot , kdeplot, rugplot
tips
```

```python
# ploting unvairate histogram
sns.histplot(tips , x = "total_bill")
```

```python
sns.displot(tips, x = "total_bill" , kind =  "hist")
```

```python
# bins parameter
sns.displot(tips, x  ="total_bill" , kind = "hist" , bins  =20)
```

```python
# It's also possible to visualize the distribution of a categorical variable using the logic of a histogram.
# Discrete bins are automatically set for categorical variables

# countplot rather than saying hist
sns.displot(tips , x  ="day" , kind  = "hist")
```

```python
# hue paramerter 
sns.displot(tips, x  = "tip" , kind  = "hist" , hue = "sex" , element  = "step")
```

```python
titanic  = sns.load_dataset("titanic")
titanic
```

```python
sns.displot(titanic , x = "age" , element  = "step" , kind  ="hist" , hue  = "sex")
```

```python
# facet using row and col # it is used for figure level function only
sns.displot(tips ,  x  ="tip" , kind = "hist" , element = "step" , col = "sex")
```

```python
#kdeplot (kernel density estimation )
# Rather than usinf discrete bins, a KDE plot smooths the observation with Gaussian kernel , producing a  continuous density  estimate 
sns.kdeplot(tips, x = "total_bill")
```

```python
sns.displot(tips , x = 'total_bill' , kind = "kde")
```

```python
# hue -> fill  If we use fill  = True then it fill the area
sns.displot(tips , x = "total_bill" , kind = "kde" , hue  = "sex" , fill = True)
```

```python
# Rugplot 
# PLot marginal distributions by drawing  ticks along the x and y axis
# This function is intended to complement other plots by showing the location of individual observation in sn unobstructive way

sns.kdeplot(tips , x  = "total_bill")
sns.rugplot(tips, x = "total_bill")
```

```python
# Bivariate histogram 
# A bivairate histogram bins the data within rectangles that title the plot
# and then shows the count of observations within each rectangle with fil color

sns.histplot(tips , x  = "total_bill" , y = "tip")
sns.displot(tips , x = "total_bill" , y  = "tip" , kind = "hist")
```

```python
# Bivariate KDEPLOT
# a bivariate KDE plot smoothes the (x,y) observations with a 2D Gausssian
sns.kdeplot(tips , x = "total_bill" , y  = "tip")
```

# MATRIX PLOTS
1. Heatmap
2. Clustermap

```python
# Heatmap

# PLot rectangular data as a  color encoded matrix
# pivot convert long data to wide format


tempt_df = gap.pivot(index = "country" , columns = "year" , values = "lifeExp")

# axes  level function
plt.figure(figsize  = (15,15))
sns.heatmap(tempt_df)
```

```python
# annot
tempt_df =  gap[gap["continent"] == "Europe"]
numeric_value = tempt_df.select_dtypes(include = "number")
```
