```python
import pandas as pd
df = pd.DataFrame()
```

to_csv


```python
df = pd.read_csv("deliveries.csv")
df.head()
```

```python
temp_df  = df.groupby("batsman")["batsman_runs"].sum().reset_index()
temp_df.to_csv("batsman_run.csv" , index = False) # index = False karne se extra index remove ho gya
```

```python
df.pivot_table(index = "batsman" , columns = "bowling_team" , values = "batsman_runs" , aggfunc = "sum")
```

to_excel

```python
temp_df  = df.groupby("batsman")["batsman_runs"].sum().reset_index()
```

```python
temp_df.to_excel("batsman.xlsx")
```

```python
temp_df.to_excel("batsman.xlsx" , sheet_name = "batsman_runs")
```

```python
temp_df2 = df.pivot_table(index = "batsman" , columns = "bowling_team" , values = 'batsman_runs' , aggfunc = "sum")
```

```python
with pd.ExcelWriter("output.xlsx") as writer :
    temp_df.to_excel(writer , sheet_name = "Sheet_name_1")
    temp_df2.to_excel(writer , sheet_name = "Sheet_name_2")
```
